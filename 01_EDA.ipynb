```python
import re

def cleanName(x):

    if pd.isna(x):
        return ""

    x = x.lower()

    x = re.sub(
        r"\b(inc|inc\.|llc|ltd|limited|corp|corporation|co|company|pvt|private)\b",
        "",
        x
    )

    x = re.sub(r"[^a-z0-9 ]", " ", x)

    x = re.sub(r"\s+", " ", x).strip()

    return x
```

Text Cleaning

```python
import re
import string

def cleanText(text):
    if pd.isna(text):
        return ""

    text = str(text).lower()

    text = re.sub(r"&", " and ", text)

    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
examples = [
    "ABC Pvt. Ltd.",
    "ABC PRIVATE LIMITED",
    "A&B Corporation",
    "Müller Group",
    "Road Street Avenue",
    None
]

for e in examples:
    print(e, " --> ", cleanText(e))
```

Feature Engineering (Ground Truth Analysis)

```python
groundTruth["matchedList"] = (
    groundTruth["matched_entity_ids"]
    .str.split(",")
)

groundTruth.head()
```

```python
source1Dict = train1.set_index("entity_id").to_dict("index")
source2Dict = train2.set_index("entity_id").to_dict("index")
source3Dict = train3.set_index("entity_id").to_dict("index")

print("Source1:", len(source1Dict))
print("Source2:", len(source2Dict))
print("Source3:", len(source3Dict))
```

```python
gt_dict = {}

for _, row in groundTruth.iterrows():

    source_id = row["source1_entity_id"]

    if pd.isna(row["matched_entity_ids"]):
        gt_dict[source_id] = set()
    else:
        gt_dict[source_id] = set(
            str(row["matched_entity_ids"]).split(",")
        )

print("Ground Truth Size:", len(gt_dict))
```

```python
list(gt_dict.items())[:5]
```

```python
source2_ids = train2["entity_id"].tolist()
source3_ids = train3["entity_id"].tolist()

print("Source2 Candidates:", len(source2_ids))
print("Source3 Candidates:", len(source3_ids))
```

```python
import random

random.seed(42)
```

```python
NEGATIVE_PER_SOURCE = 10

pairs = []

for _, row in train1.iterrows():

    source_id = row["entity_id"]

    positives = gt_dict.get(source_id, set())

    

    for match in positives:

        pairs.append({
            "source1": source_id,
            "matchedList": match,
            "label": 1
        })

    

    negatives_added = 0

    while negatives_added < NEGATIVE_PER_SOURCE:

        if random.random() < 0.5:
            candidate = random.choice(source2_ids)
        else:
            candidate = random.choice(source3_ids)

        if candidate in positives:
            continue

        pairs.append({
            "source1": source_id,
            "matchedList": candidate,
            "label": 0
        })

        negatives_added += 1
```

```python
pairs = pd.DataFrame(pairs)

print(pairs.shape)

pairs.head()
```
