Text Cleaning

```python
import re
import string

def cleanText(text):
    if text is None:
        return ""

    text = str(text).lower()
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
from difflib import SequenceMatcher
import numpy as np
```

```python
from difflib import SequenceMatcher

def similarity(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""
    return SequenceMatcher(None, a.lower(), b.lower()).ratio()


def jaccard(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""

    s1 = set(a.lower().split())
    s2 = set(b.lower().split())

    if len(s1 | s2) == 0:
        return 0.0

    return len(s1 & s2) / len(s1 | s2)


def commonWords(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""

    return len(set(a.lower().split()) & set(b.lower().split()))


def firstWordMatch(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""

    wa = a.lower().split()
    wb = b.lower().split()

    if len(wa) == 0 or len(wb) == 0:
        return 0

    return int(wa[0] == wb[0])


def lastWordMatch(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""

    wa = a.lower().split()
    wb = b.lower().split()

    if len(wa) == 0 or len(wb) == 0:
        return 0

    return int(wa[-1] == wb[-1])


def tokenDiff(a, b):
    if not isinstance(a, str):
        a = ""
    if not isinstance(b, str):
        b = ""

    return abs(len(a.split()) - len(b.split()))
```

```python
pairs["matchedName"] = pairs["matchedList"].apply(getMatchedName)

pairs["matchedAddress"] = pairs["matchedList"].apply(getMatchedAddress)

pairs["source1Name"] = pairs["source1"].apply(getSource1Name)

pairs["source1Address"] = pairs["source1"].apply(getSource1Address)
```

```python
gt_dict = dict(
    zip(
        ground_truth["source1_entity_id"],
        ground_truth["matched_entity_ids"]
    )
)
```

```python
gt_dict = {}

for _, row in ground_truth.iterrows():
    source = row["source1_entity_id"]
    matches = row["matched_entity_ids"]

    if pd.isna(matches):
        gt_dict[source] = set()
    else:
        gt_dict[source] = set(str(matches).split(","))
```

```python
def is_match(row):
    return int(
        row["matchedList"] in gt_dict.get(row["source1"], set())
    )
```

```python
pairs["label"] = pairs.apply(is_match, axis=1)
```
